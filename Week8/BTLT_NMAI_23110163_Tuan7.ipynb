#### 1. Luật Modus Ponens
$$ (P \to Q) \land P \to Q $$
* Ta có mệnh đề kéo theo có thể viết lại dưới dạng:
$$ P \to Q \equiv \neg P \lor Q $$
* Do đó, vế trái của Modus Ponens có thể viết lại thành:
$$ (\neg P \lor Q) \land P $$
* Khi này, ta có thể viết lại vế trái của Modus Ponens như sau:
$$ (\neg P \lor Q) \land (P \lor False) $$
* Áp dụng luật phân giải cho hai mệnh đề trên:
  * Mệnh đề 1 chứa: $ (\neg P \lor Q)$
  * Mệnh đề 2 chứa: $(P \lor False)$
  * Ta triệt tiêu $P$ và $\neg P$.
  * Kết quả thu được: $Q \lor False$ hay $Q$ (đpcm).




#### 2. Luật Modus Tollens
$$ (P \to Q) \land \neg Q \to \neg P $$
* Ta có mệnh đề kéo theo có thể viết lại dưới dạng:
$$ P \to Q \equiv \neg P \lor Q $$
* Do đó, vế trái của Modus Tollens có thể viết lại thành:
$$ (\neg P \lor Q) \land \neg Q $$
* Khi này, ta có thể viết lại vế trái của Modus Tollens như sau
$$ (\neg P \lor Q) \land (False \lor \neg Q) $$
* Áp dụng luật phân giải cho hai mệnh đề trên:
  * Mệnh đề 1 chứa: $ (\neg P \lor Q)$
  * Mệnh đề 2 chứa: $(False \lor \neg Q)$
  * Ta triệt tiêu $Q$ và $\neg Q$.
  * Kết quả thu được: $\neg P \lor False$ hay $\neg P$ (đpcm)

#### 3. Luật bắc cầu
$$ (P \to Q) \land (Q \to R) \to (P \to R) $$
* Ta có mệnh đề kéo theo có thể viết lại dưới dạng:
$$ P \to Q \equiv \neg P \lor Q $$
$$ Q \to R \equiv \neg Q \lor R $$
* Do đó, vế trái của luật Bắc cầu có thể viết lại thành:
$$ (\neg P \lor Q) \land (\neg Q \lor R) $$
* Áp dụng luật phân giải cho hai mệnh đề trên:
  * Mệnh đề 1 chứa: $ (\neg P \lor Q)$
  * Mệnh đề 2 chứa: $(\neg Q \lor R)$
  * Ta triệt tiêu $Q$ và $\neg Q$.
  * Kết quả thu được: $\neg P \lor R$
    * Mà theo luật kéo theo, $\neg P \lor R \equiv P \to R$ (đpcm).
