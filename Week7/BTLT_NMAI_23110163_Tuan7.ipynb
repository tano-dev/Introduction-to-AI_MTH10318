# 1. Sử dụng logic vị từ để biểu diễn các phát biểu sau:
+ a. Dê là động vật ăn cỏ.
+ b. Chó sói là động vật hung dữ.
+ c. Động vật hung dữ là động vật ăn thịt.
+ d. Động vật ăn thịt thì ăn thịt.
+ e. Động vật ăn cỏ thì ăn cỏ.
+ f. Động vật ăn thịt thì ăn các động vật ăn cỏ.
+ g. Động vật ăn thịt và động vật ăn cỏ đều uống nước.
+ h. Một động vật tiêu thụ cái mà nó uống hoặc cái mà nó ăn.



Trước hết , ta định nghĩa các ký hiệu sau:
| Vị từ                                        | Ý nghĩa               |
| -------------------------------------------- | --------------------- |
| animal(x)                      | x là động vật         |
| goat(x)                      | x là dê               |
| wolf(x)                      | x là chó sói          |
| herbivore(x)                      | x là động vật ăn cỏ   |
| carnivore(x)                      | x là động vật ăn thịt |
| dangerous(x)                      | x là động vật hung dữ |
| eat(x,y)                      | x ăn y                |
| drink(x,y)                      | x uống y              |
| consume(x,y)                      | x tiêu thụ y          |
### a. Dê là động vật ăn cỏ

$$\forall x \, (\text{goat}(x) \rightarrow \text{herbivore}(x))$$

* Ghi dưới dạng mệnh đề: "Với mọi x, nếu x là dê thì x là động vật ăn cỏ"*

### b. Chó sói là động vật hung dữ

$$\forall x \, (\text{wolf}(x) \rightarrow \text{dangerous}(x))$$

* Ghi dưới dạng mệnh đề: "Với mọi x, nếu x là chó sói thì x là động vật hung dữ"*

### c. Động vật hung dữ là động vật ăn thịt

$$\forall x \, (\text{dangerous}(x) \rightarrow \text{carnivore}(x))$$

* Ghi dưới dạng mệnh đề: "Với mọi x, nếu x là động vật hung dữ thì x là động vật ăn thịt"*

### d. Động vật ăn thịt thì ăn thịt

$$\forall x \, (\text{carnivore}(x) \rightarrow \text{eat}(x, \text{meat}))$$

* Ghi dưới dạng mệnh đề: "Với mọi x, nếu x là động vật ăn thịt thì x ăn thịt"*

### e. Động vật ăn cỏ thì ăn cỏ

$$\forall x \, (\text{herbivore}(x) \rightarrow \text{eat}(x, \text{grass}))$$

* Ghi dưới dạng mệnh đề: "Với mọi x, nếu x là động vật ăn cỏ thì x ăn cỏ"*

### f. Động vật ăn thịt thì ăn các động vật ăn cỏ

$$\forall x \, \forall y \, ((\text{carnivore}(x) \land \text{herbivore}(y)) \rightarrow \text{eat}(x, y))$$

* Ghi dưới dạng mệnh đề: "Với mọi x và y, nếu x là động vật ăn thịt và y là động vật ăn cỏ thì x ăn y"*

### g. Động vật ăn thịt và động vật ăn cỏ đều uống nước

$$\forall x \, ((\text{carnivore}(x) \lor \text{herbivore}(x)) \rightarrow \text{drink}(x, \text{water}))$$

* Ghi dưới dạng mệnh đề: "Với mọi x, nếu x là động vật ăn thịt hoặc x là động vật ăn cỏ thì x uống nước"*

### h. Một động vật tiêu thụ cái mà nó uống hoặc cái mà nó ăn

$$\forall x \, \forall y \, ((\text{animal}(x) \land (\text{eat}(x, y) \lor \text{drink}(x, y))) \rightarrow \text{consume}(x, y))$$

* Ghi dưới dạng mệnh đề "Với mọi x và y, nếu x là động vật và (x ăn y hoặc x uống y) thì x tiêu thụ y"*
